```
import pandas as pd
from nltk.corpus import wordnet as wn
from nltk.stem import WordNetLemmatizer

# Setup
lemmatizer = WordNetLemmatizer()

pos_map = {
    'NN': 'n', 'NNS': 'n', 'NNP': 'n', 'NNPS': 'n',
    'VB': 'v', 'VBD': 'v', 'VBG': 'v', 'VBN': 'v', 'VBP': 'v', 'VBZ': 'v',
    'JJ': 'a', 'JJR': 'a', 'JJS': 'a',
    'RB': 'r', 'RBR': 'r', 'RBS': 'r'
}

IRREGULAR_LEMMAS = {
    'best': 'good',    'better': 'good',
    'worst': 'bad',    'worse': 'bad',
    'further': 'far',  'furthest': 'far',
    'least': 'little', 'less': 'little',
}

# Conversion function
def get_sense_key(lemma, prediction, pos=None):
    try:
        # MFS predictions are already sense keys (contain %) — pass through directly
        if "%" in str(prediction):
            return prediction

        synset = wn.synset(prediction)
        synset_lemmas = {l.name().lower(): l for l in synset.lemmas()}

        # Strategy 1: exact UFSAC lemma match
        if str(lemma).lower() in synset_lemmas:
            return synset_lemmas[str(lemma).lower()].key()

        # Strategy 2: NLTK lemmatizer fallback
        wn_pos = pos_map.get(pos, synset.pos())
        nltk_lemma = lemmatizer.lemmatize(str(lemma).lower(), pos=wn_pos)
        if nltk_lemma in synset_lemmas:
            return synset_lemmas[nltk_lemma].key()

        # Strategy 3: Irregular lemma mapping
        irregular = IRREGULAR_LEMMAS.get(str(lemma).lower())
        if irregular and irregular in synset_lemmas:
            return synset_lemmas[irregular].key()

        return ""
    except:
        return ""

# Load the fixed MFS file
df = pd.read_csv("/content/drive/MyDrive/EWISER/ewiser_full_mfs.csv")
print("Loaded:", df.shape)
print("Verify first row — target_word should match lemma:")
print(df[['target_word', 'lemma', 'pos', 'final_prediction']].head(3))

# Apply conversion
df["predicted_sense_key"] = df.apply(
    lambda row: get_sense_key(
        row["lemma"],
        row["final_prediction"],
        row["pos"]
    )
    if pd.notna(row["final_prediction"]) and row["final_prediction"] != ""
    else "",
    axis=1
)

# Stats
total          = len(df)
has_key        = (df["predicted_sense_key"] != "") & (df["predicted_sense_key"].notna())
failed         = (~has_key).sum()
correct        = (df["predicted_sense_key"] == df["gold_sense"]).sum()

print("RESULTS")
print(f"Total instances:      {total:,}")
print(f"Sense keys generated: {has_key.sum():,}")
print(f"Failed conversions:   {failed:,}")
print(f"Correct predictions:  {correct:,}")
print(f"Accuracy:             {correct/total*100:.2f}%")
print(f"Coverage:             {has_key.sum()/total*100:.2f}%")

# Save
output_file = "/content/drive/MyDrive/EWISER/ewiser_full_mfs_with_sensekey.csv"
df.to_csv(output_file, index=False)
print("Saved:", output_file)
```

```
import pandas as pd
import time
df = pd.read_csv(
    "/content/drive/MyDrive/EWISER/semcor_fulldataset.csv"
)

START = 0
END = 5000         #change accordingly

chunk = df.iloc[START:END]

print("Rows to process:", len(chunk))

results = []

start_time = time.time()

# CACHE PREVIOUS SENTENCE
prev_sentence = None
prev_doc = None

# PROCESS

for idx, row in chunk.iterrows():

    sentence = str(row["sentence"])
    target_word = str(row["target_word"])

    try:

        # Only run EWISER once per sentence
        if sentence != prev_sentence:

            prev_doc = wsd(nlp(sentence))
            prev_sentence = sentence

        doc = prev_doc

        prediction = ""

        for token in doc:

            if token.text == target_word and token._.synset:

                prediction = token._.synset.name()
                break

        results.append([
            sentence,
            target_word,
            row["lemma"],
            row["pos"],
            row["gold_sense"],
            prediction
        ])

    except Exception as e:

        print(f"Error at row {idx}")
        print(e)

        results.append([
            sentence,
            target_word,
            row["lemma"],
            row["pos"],
            row["gold_sense"],
            ""
        ])

    # PROGRESS
    processed = idx - START

    if processed % 10 == 0:
        print(f"Processed: {processed}")

    # CHECKPOINT EVERY 1000
    if processed > 0 and processed % 1000 == 0:

        checkpoint_file = (
            f"/content/drive/MyDrive/EWISER/"
            f"checkpoint_{START}_{END}.csv"
        )

        pd.DataFrame(
            results,
            columns=[
                "sentence",
                "target_word",
                "lemma",
                "pos",
                "gold_sense",
                "ewiser_prediction"
            ]
        ).to_csv(
            checkpoint_file,
            index=False
        )

        print(
            f"Checkpoint saved ({processed} rows)"
        )

# FINAL SAVE
results_df = pd.DataFrame(
    results,
    columns=[
        "sentence",
        "target_word",
        "lemma",
        "pos",
        "gold_sense",
        "ewiser_prediction"
    ]
)

output_file = (
    f"/content/drive/MyDrive/EWISER/"
    f"ewiser_{START}_{END}.csv"
)

results_df.to_csv(
    output_file,
    index=False
)

# COVERAGE
predicted = (
    results_df["ewiser_prediction"]
    .fillna("")
    .ne("")
    .sum()
)

coverage = predicted / len(results_df)

# TIMING
end_time = time.time()

total_time = end_time - start_time

print("Completed Successfully")

print("Rows Processed :", len(results_df))
print("Predicted Rows :", predicted)
print("Coverage       :", round(coverage * 100, 2), "%")

print("Time Seconds   :", round(total_time, 2))
print("Time Minutes   :", round(total_time / 60, 2))

print("Saved File     :", output_file)
```

```
# Step 7: Evaluate accuracy
# EWISER predictions - These are synsets:state.v.01
# MFS predictions - These are sense keys:say%2:32:00::

from nltk.corpus import wordnet as wn

correct = 0
predicted = 0

for i, row in df.iterrows():

    pred = row["final_prediction"]

    if pd.isna(pred) or pred == "":
        continue

    try:

        gold_syn = wn.lemma_from_key(
            row["gold_sense"]
        ).synset()

        if "." in str(pred):

            pred_syn = wn.synset(pred)

        else:

            pred_syn = wn.lemma_from_key(
                pred
            ).synset()

        predicted += 1

        if gold_syn == pred_syn:
            correct += 1

    except:
        pass

accuracy = correct / predicted

print("Predicted:", predicted)
print("Correct:", correct)
print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)
```

```
# TOP MISCLASSIFIED LEMMAS FOR:
# EWISER (No MFS)
# EWISER + MFS

import pandas as pd

# CHANGE THIS ONLY
TOP_N = 20

# FILE PATHS
ewiser_nomfs_path = "/content/drive/MyDrive/EWISER/ewiser_full_with_sensekey.csv"
ewiser_mfs_path   = "/content/drive/MyDrive/EWISER/ewiser_full_mfs_with_sensekey.csv"

# HELPER FUNCTION
def analyze_model(file_path, model_name):
    """
    Reads a model result CSV and prints:
    1) Top-N most misclassified lemmas by error count
    2) Top-N highest error-rate lemmas (with min frequency filter)
    """
    print(f"{model_name} - MISCLASSIFIED LEMMA ANALYSIS")

    df = pd.read_csv(file_path)

    # Basic cleanup
    # Remove rows where lemma is missing
    df = df[df["lemma"].notna()].copy()

    # Convert lemma to string so grouping is safe
    df["lemma"] = df["lemma"].astype(str)

    # Mark correctness
    # Assumes these columns exist:
    # gold_sense
    # predicted_sense_key
    df["correct"] = df["predicted_sense_key"] == df["gold_sense"]

    # Separate wrong rows
    errors = df[~df["correct"]].copy()

    # A) TOP-N BY RAW ERROR COUNT
    print(f"\nA. Top {TOP_N} most misclassified lemmas by ERROR COUNT\n")

    # Count how many errors each lemma has
    top_by_count = (
        errors.groupby("lemma")
        .size()
        .sort_values(ascending=False)
        .head(20)
    )

    print(f"Number of lemmas shown: {len(top_by_count)}\n")

    for lemma, error_count in top_by_count.items():
        total_count = (df["lemma"] == lemma).sum()
        error_rate = (error_count / total_count) * 100 if total_count > 0 else 0

        print(
            f"{lemma:<20} "
            f"errors: {error_count:>5}  /  "
            f"total: {total_count:>6}  "
            f"({error_rate:.1f}% error rate)"
        )

    # B) TOP-N BY ERROR RATE
    print(f"\n\nB. Top {TOP_N} most misclassified lemmas by ERROR RATE\n")

    # First compute total occurrences per lemma
    total_per_lemma = df.groupby("lemma").size().rename("total_count")

    # Then compute error count per lemma
    error_per_lemma = errors.groupby("lemma").size().rename("error_count")

    # Merge both
    summary = pd.concat([total_per_lemma, error_per_lemma], axis=1).fillna(0)

    summary["error_count"] = summary["error_count"].astype(int)

    # Error rate
    summary["error_rate"] = summary["error_count"] / summary["total_count"] * 100

    # To avoid silly cases where a lemma appears only once
    # and gets 100% error rate, apply a minimum frequency filter.
    # You can change this threshold if you want.
    MIN_FREQ = 10

    summary_rate = summary[summary["total_count"] >= MIN_FREQ].copy()

    top_by_rate = (
        summary_rate.sort_values(
            by=["error_rate", "error_count"],
            ascending=[False, False]
        )
        .head(TOP_N)
    )

    print(f"Minimum frequency filter used for error-rate ranking: {MIN_FREQ}")
    print(f"Number of lemmas shown: {len(top_by_rate)}\n")

    for lemma, row in top_by_rate.iterrows():
        print(
            f"{lemma:<20} "
            f"errors: {int(row['error_count']):>5}  /  "
            f"total: {int(row['total_count']):>6}  "
            f"({row['error_rate']:.1f}% error rate)"
        )

# EWISER (No MFS)
analyze_model(ewiser_nomfs_path, "EWISER (No MFS)")

# EWISER + MFS
analyze_model(ewiser_mfs_path, "EWISER + MFS")
```

```
# POS Filtering
import pandas as pd
import re

# Load the EWISER MFS file
df = pd.read_csv("/content/drive/MyDrive/EWISER/ewiser_full_mfs_with_sensekey.csv")
df['correct'] = df['predicted_sense_key'] == df['gold_sense']

# Map Penn Treebank POS tags (in SemCor) to WordNet POS codes (n/v/a/r)
pos_map = {
    'NN': 'n', 'NNS': 'n', 'NNP': 'n', 'NNPS': 'n',          # nouns
    'VB': 'v', 'VBD': 'v', 'VBG': 'v', 'VBN': 'v',             # verbs
    'VBP': 'v', 'VBZ': 'v',
    'JJ': 'a', 'JJR': 'a', 'JJS': 'a',                          # adjectives
    'RB': 'r', 'RBR': 'r', 'RBS': 'r'                           # adverbs
}

def get_pred_wn_pos(sense_key):
    """
    Extract the WordNet POS from a predicted sense key.
    Sense key format: lemma%ss_type:lex_filenum:lex_id
    ss_type: 1=noun, 2=verb, 3=adj, 4=adv, 5=adj satellite
    """
    if pd.isna(sense_key) or sense_key == '':
        return None
    m = re.match(r'^[^%]+%(\d):', str(sense_key))
    if m:
        return {'1':'n', '2':'v', '3':'a', '4':'r', '5':'a'}.get(m.group(1))
    return None

# Get the gold POS from SemCor (this is reliable — it's from the annotation)
df['gold_wn_pos'] = df['pos'].map(pos_map)

# Get the predicted POS from EWISER's output sense key
df['pred_wn_pos'] = df['predicted_sense_key'].apply(get_pred_wn_pos)

# Find cross-POS errors: rows where prediction has WRONG part of speech
# e.g., gold is noun (n) but EWISER predicted a verb sense (v)
cross_pos_errors = df[
    df['predicted_sense_key'].notna() &           # must have a prediction
    (df['predicted_sense_key'] != '') &            # must not be empty
    (df['gold_wn_pos'] != df['pred_wn_pos']) &    # POS must DIFFER
    df['gold_wn_pos'].notna() &                   # gold must have a POS
    df['pred_wn_pos'].notna()                     # prediction must have a POS
]

print(f"Total instances: {len(df):,}")
print(f"Cross-POS errors: {len(cross_pos_errors):,} ({len(cross_pos_errors)/len(df)*100:.1f}%)")
print(f"These errors could be avoided if EWISER's candidate generator used SemCor's gold POS.")

print("\nSample cross-POS errors:")
print(cross_pos_errors[['target_word','pos','gold_wn_pos','pred_wn_pos',
                         'gold_sense','predicted_sense_key']].head(10).to_string())

print("\nCross-POS errors by POS tag:")
print(cross_pos_errors.groupby(['pos','gold_wn_pos','pred_wn_pos']).size()
      .sort_values(ascending=False).head(10))
```

```
# Axis 1 — Error Type Distribution per Word
import pandas as pd, re

ew = pd.read_csv("/content/drive/MyDrive/EWISER/ewiser_full_mfs_with_sensekey.csv")
gb = pd.read_csv("/content/drive/MyDrive/EWISER/glossbert_full.csv")
ew['correct'] = ew['predicted_sense_key'] == ew['gold_sense']
gb['correct'] = gb['glossbert_prediction'] == gb['gold_sense']

def get_lex(sk):
    if pd.isna(sk) or sk == '': return None
    sk = str(sk).split(';')[0]
    m = re.match(r'[^%]+%(\d):(\d{2})', sk)
    return m.group(2) if m else None

GROUP_A_LEMMAS = ['make', 'do', 'find', 'give', 'see', 'be', 'have']

print(f"{'Word':<8} {'GB errors':>10} {'EW errors':>10} "
      f"{'GB same-lex%':>14} {'EW same-lex%':>14}")
print("-" * 60)

for lemma in GROUP_A_LEMMAS:
    # GlossBERT
    gb_w = gb[(gb['target_word'] == lemma) & (~gb['correct'])].copy()
    gb_w['gl'] = gb_w['gold_sense'].apply(get_lex)
    gb_w['pl'] = gb_w['glossbert_prediction'].apply(get_lex)
    gb_same = (gb_w['gl'] == gb_w['pl']).sum()
    gb_pct = gb_same / len(gb_w) * 100 if len(gb_w) > 0 else 0

    # EWISER
    ew_w = ew[(ew['lemma'] == lemma) & (~ew['correct'])].copy()
    ew_w['gl'] = ew_w['gold_sense'].apply(get_lex)
    ew_w['pl'] = ew_w['predicted_sense_key'].apply(get_lex)
    ew_same = (ew_w['gl'] == ew_w['pl']).sum()
    ew_pct = ew_same / len(ew_w) * 100 if len(ew_w) > 0 else 0

    print(f"{lemma:<8} {len(gb_w):>10,} {len(ew_w):>10,} "
          f"{gb_pct:>13.1f}% {ew_pct:>13.1f}%")
```

```
import pandas as pd
import re
import os

DRIVE_FOLDER = '/content/drive/MyDrive/EWISER'  #  change if needed

EW_FILE  = os.path.join(DRIVE_FOLDER, 'ewiser_full_mfs_with_sensekey.csv')
OUT_FILE = os.path.join(DRIVE_FOLDER, 'ewiser_pos_filtered.csv')

print("Loading EWISER predictions...")
ew = pd.read_csv(EW_FILE)
print(f"  Rows loaded: {len(ew):,}")
print(f"  Columns: {list(ew.columns)}")

# Map POS and find cross-POS errors
# Step A: SemCor uses Penn Treebank POS tags like NN, VBD, JJ, RB
# We convert these to simple 4-letter codes: n, v, a, r
penn_to_simple = {
    'NN':'n',  'NNS':'n',  'NNP':'n',  'NNPS':'n',   # nouns
    'VB':'v',  'VBD':'v',  'VBZ':'v',  'VBP':'v',     # verbs
    'VBG':'v', 'VBN':'v',                              # verbs (cont.)
    'JJ':'a',  'JJR':'a',  'JJS':'a',                 # adjectives
    'RB':'r',  'RBR':'r',  'RBS':'r'                   # adverbs
}
ew['gold_pos'] = ew['pos'].map(penn_to_simple)

# Step B: Extract POS from the predicted sense key
# WordNet sense key format: lemma%ss_type:lex_filenum:lex_id::
# ss_type digit tells us the POS:
#   1 = noun, 2 = verb, 3 = adjective, 4 = adverb, 5 = adjective satellite
def get_pos_from_sense_key(sense_key):
    m = re.match(r'[^%]+%(\d):', str(sense_key))
    if m:
        ss_type = m.group(1)
        return {'1':'n', '2':'v', '3':'a', '4':'r', '5':'a'}.get(ss_type)
    return None  # could not parse

ew['pred_pos'] = ew['predicted_sense_key'].apply(get_pos_from_sense_key)

# Step C: Mark each row as correct or wrong (before any fix)
ew['correct_before'] = (ew['predicted_sense_key'] == ew['gold_sense'])
total_before = ew['correct_before'].sum()
acc_before   = total_before / len(ew) * 100
print(f"Accuracy BEFORE filter: {acc_before:.2f}%  ({total_before:,} correct)")

# Step D: Find cross-POS errors
#   A cross-POS error = prediction was WRONG AND predicted POS ≠ gold POS
cross_pos_mask = (
    ~ew['correct_before'] &        # wrong prediction
    ew['gold_pos'].notna()  &      # gold POS is known
    ew['pred_pos'].notna()  &      # predicted POS is known
    (ew['gold_pos'] != ew['pred_pos'])  # POS mismatch!
)
n_cross = cross_pos_mask.sum()
print(f"Cross-POS errors found: {n_cross:,}  ({n_cross/len(ew)*100:.1f}% of all instances)")

# Show some examples to understand what these look like
print("\nExamples of cross-POS errors:")
for i, r in ew[cross_pos_mask].head(5).iterrows():
    print(f"  word={r['lemma']:<12}  "
          f"SemCor tag={r['pos']:<5} (gold_pos={r['gold_pos']})  "
          f"predicted pos={r['pred_pos']}")
    print(f"    gold_sense:      {r['gold_sense']}")
    print(f"    predicted_sense: {r['predicted_sense_key']}")
    print()

# Build MFS dictionary and apply the fix

# Step E: Build MFS (Most Frequent Sense) dictionary
# For each (lemma, POS) pair, find the most common gold annotation
# in the SemCor corpus. This is our replacement sense.
mfs_dict = {}
for (lemma, pos), group in ew.groupby(['lemma', 'gold_pos']):
    freq = group['gold_sense'].value_counts()
    if len(freq) > 0:
        mfs_dict[(lemma, pos)] = freq.index[0]  # most frequent gold sense

print(f"MFS dictionary built: {len(mfs_dict):,} (lemma, POS) combinations")

# Step F: Apply the fix
# For each cross-POS error: replace predicted sense with MFS for correct POS
ew_fixed = ew.copy()
fix_count = 0

for idx in ew[cross_pos_mask].index:
    lemma    = ew.at[idx, 'lemma']
    gold_pos = ew.at[idx, 'gold_pos']
    lookup_key = (lemma, gold_pos)
    replacement = mfs_dict.get(lookup_key)
    if replacement:
        ew_fixed.at[idx, 'predicted_sense_key'] = replacement
        fix_count += 1

print(f"Replacements applied: {fix_count:,}")

# Compute new accuracy and save output
ew_fixed['correct_after'] = (ew_fixed['predicted_sense_key'] == ew_fixed['gold_sense'])
total_after = ew_fixed['correct_after'].sum()
acc_after   = total_after / len(ew_fixed) * 100
new_correct = total_after - total_before

print("RESULTS — POS Filtering Impact")
print(f"  Accuracy BEFORE: {acc_before:.2f}%  ({total_before:,} correct)")
print(f"  Accuracy AFTER:  {acc_after:.2f}%  ({total_after:,} correct)")
print(f"  Improvement:     +{acc_after - acc_before:.2f} percentage points")
print(f"  New correct:     +{new_correct:,} predictions")
print(f"  Fixed -> correct: {new_correct:,} / {fix_count:,} = {new_correct/fix_count*100:.0f}%")
print(f"  Fixed -> wrong:   {fix_count - new_correct:,} (MFS was not the gold sense)")

# Save output file
output = ew_fixed[[
    'sentence', 'target_word', 'lemma', 'pos',
    'gold_sense', 'predicted_sense_key',
    'correct_before', 'correct_after'
]].copy()
output['pos_filter_applied'] = cross_pos_mask.map(
    {True: 'YES - replaced with MFS', False: 'no'})

output.to_csv(OUT_FILE, index=False)
print(f"\nSaved to: {OUT_FILE}")
print(f"  Rows: {len(output):,}")
print(f"  Rows where fix was applied: {(output['pos_filter_applied']=='YES - replaced with MFS').sum():,}")

# Show sample fixed rows
print("\nSample rows where fix was applied:")
fixed_sample = output[output['pos_filter_applied'] == 'YES - replaced with MFS'].head(8)
for i, r in fixed_sample.iterrows():
    print(f"  [{r['lemma']:<12}] gold={r['gold_sense'][:35]:<35} "
          f"after_fix={r['predicted_sense_key'][:35]:<35} correct={r['correct_after']}")
```

```
import pandas as pd
import re
import os

DRIVE_FOLDER = '/content/drive/MyDrive/EWISER'  # change if needed

EW_FILE  = os.path.join(DRIVE_FOLDER, 'ewiser_full_mfs_with_sensekey.csv')
OUT_FILE = os.path.join(DRIVE_FOLDER, 'ewiser_pos_filtered.csv')

print("Loading EWISER predictions...")
ew = pd.read_csv(EW_FILE)
print(f"  Rows loaded: {len(ew):,}")
print(f"  Columns: {list(ew.columns)}")


# Load prediction files
print("Loading prediction files...")
gb = pd.read_csv(GB_FILE)
ew = pd.read_csv(EW_FILE)
gb['correct'] = gb['glossbert_prediction'] == gb['gold_sense']
ew['correct'] = ew['predicted_sense_key']  == ew['gold_sense']
print(f"  GlossBERT: {len(gb):,} rows")
print(f"  EWISER:    {len(ew):,} rows")

def get_lex(sk):
    """Extract lexicographer file number from sense key.
    Format: lemma%ss_type:lex_filenum:lex_id::
    e.g. man%1:18:04:: → lex_filenum = '18'
    """
    m = re.match(r'[^%]+%\d:(\d{2}):', str(sk))
    return m.group(1) if m else None

# WORD 1: MAN
print("WORD: MAN")

gb_man = gb[gb['target_word'] == 'man'].copy()
ew_man = ew[ew['lemma'] == 'man'].copy()

print(f"\nInstances:  GB={len(gb_man):,}  EW={len(ew_man):,}")
print(f"Errors:     GB={((~gb_man['correct']).sum()):,} ({(~gb_man['correct']).mean()*100:.1f}%)"
      f"  EW={((~ew_man['correct']).sum()):,} ({(~ew_man['correct']).mean()*100:.1f}%)")

print("\nGlossBERT top error patterns:")
gb_err = gb_man[~gb_man['correct']]
top = gb_err.groupby(['gold_sense','glossbert_prediction']).size().reset_index(name='n').nlargest(5,'n')
for i, r in top.iterrows():
    print(f"  gold={r['gold_sense'][:30]:<30} pred={r['glossbert_prediction'][:30]:<30} n={r['n']}")

print("\nEWISER top error patterns:")
ew_err = ew_man[~ew_man['correct']]
top2 = ew_err.groupby(['gold_sense','predicted_sense_key']).size().reset_index(name='n').nlargest(5,'n')
for i, r in top2.iterrows():
    print(f"  gold={r['gold_sense'][:30]:<30} pred={r['predicted_sense_key'][:30]:<30} n={r['n']}")

# FLOWCHART - applied to classify errors
# Built from reading OEWN 2025 sense definitions for "man"
SENSES_MAN = {
    'man%1:05:00': 'adult male person (individual)',
    'man%1:05:01': 'male person in a specific role (medicine man, best man)',
    'man%1:18:00': 'humanity / the human race',
    'man%1:18:03': 'male who is a fellow group member (soldiers, team)',
    'man%1:18:04': 'male subordinate / employee'
}
GENERIC_PATTERNS = ['contemporary man', 'modern man', 'biblical man', 'ancient man',
                    'god created man', "man 's", 'man has', 'man is',
                    'the nature of man', 'idea of man', 'man was made']

def apply_man_flowchart(gold, pred, sentence):
    """
    Apply the 4-question flowchart to classify a man error.
    Returns: (category, reason)
    """
    sent_l = sentence.lower()
    g_lex = get_lex(gold)
    p_lex = get_lex(pred)

    # Q3 in the flowchart: generic bare-noun use - Category C
    if any(pat in sent_l for pat in GENERIC_PATTERNS):
        if g_lex in ('05','18') and p_lex in ('05','18'):
            return ('C',
                'Q3 YES: Generic bare-noun "man" without article — simultaneously '
                'valid as individual (05:00) and as humanity (18:00). Flowchart '
                'cannot decide. This is a WordNet granularity problem.')

    # Same lex file - fine-grained Category C
    if g_lex and p_lex and g_lex == p_lex:
        return ('C',
            f'Same lex file (lex-{g_lex}) — senses too close to separate '
            f'from text. Flowchart cannot decide between sub-senses.')

    # Cross-lex error - Category A (genuine model error)
    return ('A',
        f'Cross-lex error: gold lex-{g_lex} vs predicted lex-{p_lex}. '
        f'Model predicted wrong semantic domain.')

print("\nApplying flowchart to classify errors...")
gb_man_err = gb_man[~gb_man['correct']].copy()
ew_man_err = ew_man[~ew_man['correct']].copy()

gb_man_err[['category','reason']] = pd.DataFrame(
    gb_man_err.apply(
        lambda r: apply_man_flowchart(r['gold_sense'], r['glossbert_prediction'], r['sentence']),
        axis=1).tolist(), index=gb_man_err.index)

ew_man_err[['category','reason']] = pd.DataFrame(
    ew_man_err.apply(
        lambda r: apply_man_flowchart(r['gold_sense'], r['predicted_sense_key'], r['sentence']),
        axis=1).tolist(), index=ew_man_err.index)

print(f"\nMAN classification results:")
print(f"  GlossBERT: {gb_man_err['category'].value_counts().to_dict()}")
print(f"  EWISER:    {ew_man_err['category'].value_counts().to_dict()}")

#WORD 2: UNITED_STATES
print("WORD: UNITED_STATES")

ew_us = ew[ew['lemma'] == 'united_states'].copy()
gb_us = gb[gb['target_word'] == 'united_states'].copy()
print(f"GlossBERT: {len(gb_us)} instances  "
      f"({(~gb_us['correct']).sum()} errors = {(~gb_us['correct']).mean()*100:.0f}%)")
print(f"EWISER:    {len(ew_us):,} instances  "
      f"({(~ew_us['correct']).sum()} errors = {(~ew_us['correct']).mean()*100:.0f}%)")

GOV_SIGNALS = ['appropriated','arms policy','nuclear','congress','senate',
               'military','enacted','legislation','defense','foreign policy']

def apply_us_flowchart(gold, pred, sentence):
    sl = sentence.lower()
    is_gov = any(s in sl for s in GOV_SIGNALS)
    if is_gov:
        return ('B',
            'Government action detected (Q2 YES): sentence describes '
            'institutional/policy action. United States is acting as '
            'government (15:00), not as country (14:00). EWISER correct. '
            'Gold annotation wrong.')
    return ('C',
        'Country vs government indistinguishable (Q2 NO → Q3 CANNOT DECIDE): '
        'sentence refers to the US as a political entity but neither '
        'country-territory nor government-institution is exclusively signalled.')

ew_us_err = ew_us[~ew_us['correct']].copy()
ew_us_err[['category','reason']] = pd.DataFrame(
    ew_us_err.apply(
        lambda r: apply_us_flowchart(r['gold_sense'], r['predicted_sense_key'], r['sentence']),
        axis=1).tolist(), index=ew_us_err.index)

print(f"\nUNITED_STATES EWISER errors: {ew_us_err['category'].value_counts().to_dict()}")

# WORD 3: ONE
print("WORD: ONE and TWO — ss_type encoding bug investigation")
for word in ['one','two']:
    gb_w = gb[gb['target_word']==word].copy()
    ew_w = ew[ew['lemma']==word].copy()
    gb_w['correct'] = gb_w['glossbert_prediction'] == gb_w['gold_sense']
    ew_w['correct'] = ew_w['predicted_sense_key']  == ew_w['gold_sense']
    gb_err = gb_w[~gb_w['correct']]
    bug_count = ((gb_err['gold_sense'].str.contains('%3:',na=False)) &
                 (gb_err['glossbert_prediction'].str.contains('%5:',na=False))).sum()
    print(f"\n{word.upper()}: GB errors={len(gb_err)} | ss_type bug={bug_count} "
          f"({bug_count/len(gb_err)*100:.0f}%) | genuine={len(gb_err)-bug_count}")
    print(f"        EW errors={(~ew_w['correct']).sum()} | all genuine")
```

```
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import re
import os

DRIVE_FOLDER = '/content/drive/MyDrive/EWISER'  # change if your folder is different

GB_FILE = os.path.join(DRIVE_FOLDER, 'glossbert_full.csv')
EW_FILE = os.path.join(DRIVE_FOLDER, 'ewiser_full_mfs_with_sensekey.csv')

for f in [GB_FILE, EW_FILE]:
    print(f"{'found' if os.path.exists(f) else 'NOT FOUND'} {f}")

print("\nLoading GlossBERT predictions...")
gb = pd.read_csv(GB_FILE)
gb['correct'] = gb['glossbert_prediction'] == gb['gold_sense']
print(f"  Rows: {len(gb):,}  |  Overall accuracy: {gb['correct'].mean()*100:.2f}%")

print("Loading EWISER predictions...")
ew = pd.read_csv(EW_FILE)
ew['correct'] = ew['predicted_sense_key'] == ew['gold_sense']
print(f"  Rows: {len(ew):,}  |  Overall accuracy: {ew['correct'].mean()*100:.2f}%")

# Helper: extract lex file number from a sense key
def get_lex(sk):
    m = re.match(r'[^%]+%\d:(\d{2}):', str(sk))
    return m.group(1) if m else None

print("Same-lex error % per word (light verbs)")

LIGHT_VERBS = ['make', 'do', 'find', 'give', 'see', 'be', 'have']

print(f"\n{'Word':<8} {'GB errors':>10} {'EW errors':>10} "
      f"{'GB same-lex%':>14} {'EW same-lex%':>14}")

for lemma in LIGHT_VERBS:
    # GlossBERT: filter target_word == lemma, wrong predictions only
    gb_w = gb[(gb['target_word'] == lemma) & (~gb['correct'])].copy()
    gb_w['gold_lex'] = gb_w['gold_sense'].apply(get_lex)
    gb_w['pred_lex'] = gb_w['glossbert_prediction'].apply(get_lex)
    gb_same = (gb_w['gold_lex'] == gb_w['pred_lex']).sum()
    gb_pct  = gb_same / len(gb_w) * 100 if len(gb_w) > 0 else 0

    # EWISER: filter lemma == lemma (EWISER uses lemma column, not target_word)
    ew_w = ew[(ew['lemma'] == lemma) & (~ew['correct'])].copy()
    ew_w['gold_lex'] = ew_w['gold_sense'].apply(get_lex)
    ew_w['pred_lex'] = ew_w['predicted_sense_key'].apply(get_lex)
    ew_same = (ew_w['gold_lex'] == ew_w['pred_lex']).sum()
    ew_pct  = ew_same / len(ew_w) * 100 if len(ew_w) > 0 else 0

    print(f"{lemma:<8} {len(gb_w):>10,} {len(ew_w):>10,} "
          f"{gb_pct:>13.1f}% {ew_pct:>13.1f}%")


print("Bias direction (Extended Shared Failure Analysis)")

# Light verbs + content nouns that were analysed
ALL_WORDS = ['make', 'have', 'do', 'find', 'give', 'see', 'be', 'man']

print(f"\n{'Word':<8} {'GlossBERT default wrong sense':<45} "
      f"{'EWISER default wrong sense':<45} {'Direction'}")

bias_rows = []
for word in ALL_WORDS:
    gb_w = gb[(gb['target_word'] == word) & (~gb['correct'])]
    ew_w = ew[(ew['lemma'] == word)       & (~ew['correct'])]

    gb_top = gb_w['glossbert_prediction'].value_counts()
    ew_top = ew_w['predicted_sense_key'].value_counts()

    gb_default = gb_top.index[0] if len(gb_top) > 0 else 'N/A'
    ew_default = ew_top.index[0] if len(ew_top) > 0 else 'N/A'
    gb_n       = int(gb_top.iloc[0]) if len(gb_top) > 0 else 0
    ew_n       = int(ew_top.iloc[0]) if len(ew_top) > 0 else 0
    direction  = 'SAME' if gb_default == ew_default else 'DIFFERENT'

    print(f"{word:<8} {gb_default[:40]:<45} {ew_default[:40]:<45} {direction}")
    print(f"         (n={gb_n})                                         (n={ew_n})")

    bias_rows.append({
        'word'            : word,
        'gb_default_wrong': gb_default,
        'gb_count'        : gb_n,
        'ew_default_wrong': ew_default,
        'ew_count'        : ew_n,
        'bias_direction'  : direction
    })

OUT_BIAS = os.path.join(DRIVE_FOLDER, 'shared_failure_bias_direction.csv')
pd.DataFrame(bias_rows).to_csv(OUT_BIAS, index=False)
print(f"\nSaved: {OUT_BIAS}")

print("Shared failure analysis")

# Step 1: Build composite key
gb['row_id'] = (gb['sentence'].astype(str).str.strip() + '||' +
                gb['target_word'].astype(str).str.strip() + '||' +
                gb['gold_sense'].astype(str).str.strip())

ew['row_id'] = (ew['sentence'].astype(str).str.strip() + '||' +
                ew['target_word'].astype(str).str.strip() + '||' +
                ew['gold_sense'].astype(str).str.strip())

# Step 2: Inner join — keep only rows both models evaluated
merged = pd.merge(
    ew[['row_id', 'lemma', 'correct']].rename(columns={'correct': 'ew_c'}),
    gb[['row_id', 'correct']].rename(columns={'correct': 'gb_c'}),
    on='row_id',
    how='inner'
)
print(f"\nMatched instances (same sentence, both models): {len(merged):,}")

# Step 3: Overall 4-outcome counts
both_c  = ( merged['gb_c'] &  merged['ew_c']).sum()
both_w  = (~merged['gb_c'] & ~merged['ew_c']).sum()
only_gb = (~merged['gb_c'] &  merged['ew_c']).sum()  # GB wrong, EW correct
only_ew = ( merged['gb_c'] & ~merged['ew_c']).sum()  # EW wrong, GB correct

print(f"\nOverall results:")
print(f"  Both correct:          {both_c:>7,}  ({both_c/len(merged)*100:.1f}%)")
print(f"  Both wrong:            {both_w:>7,}  ({both_w/len(merged)*100:.1f}%)")
print(f"  Only GlossBERT wrong:  {only_gb:>7,}  ({only_gb/len(merged)*100:.1f}%)")
print(f"  Only EWISER wrong:     {only_ew:>7,}  ({only_ew/len(merged)*100:.1f}%)")
print(f"  Total matched:         {len(merged):>7,}  (100%)")

# Step 4: Word-level breakdown
print(f"\n{'Word':<8} {'Total':>7} {'Both':>8} {'Both':>8} "
      f"{'Only GB':>10} {'Only EW':>10} {'GB err%':>9} {'EW err%':>9}")

rows = []
for word in ['make', 'have', 'do', 'find', 'give', 'see', 'be', 'man', 'one', 'two']:
    m = merged[merged['lemma'] == word]
    if len(m) < 5:
        continue
    bc  = ( m['gb_c'] &  m['ew_c']).sum()
    bw  = (~m['gb_c'] & ~m['ew_c']).sum()
    ogb = (~m['gb_c'] &  m['ew_c']).sum()  # GB wrong, EW correct
    oew = ( m['gb_c'] & ~m['ew_c']).sum()  # EW wrong, GB correct
    gbe = (~m['gb_c']).mean() * 100
    ewe = (~m['ew_c']).mean() * 100
    print(f"{word:<8} {len(m):>7,} {bc:>8,} {bw:>8,} {ogb:>10,} {oew:>10,} "
          f"{gbe:>8.1f}% {ewe:>8.1f}%")
    rows.append({
        'word': word, 'total': len(m),
        'both_correct': bc, 'both_wrong': bw,
        'only_gb_wrong': ogb, 'only_ew_wrong': oew,
        'gb_err_pct': round(gbe, 1), 'ew_err_pct': round(ewe, 1)
    })

OUT_WORD = os.path.join(DRIVE_FOLDER, 'shared_failure_by_word.csv')
pd.DataFrame(rows).to_csv(OUT_WORD, index=False)
print(f"\nSaved: {OUT_WORD}")


print("Content noun error rates (man, US, one, two)")

for word in ['man', 'united_states', 'one', 'two']:
    gb_word = gb[gb['target_word'] == word]
    ew_word = ew[ew['lemma'] == word]
    gb_n = len(gb_word); gb_err = (~gb_word['correct']).sum()
    ew_n = len(ew_word); ew_err = (~ew_word['correct']).sum()

    print(f"WORD: {word.upper()}")
    print(f"  GlossBERT: {gb_n:>5} instances | {gb_err:>4} errors | "
          f"{gb_err/gb_n*100:.1f}% error rate")
    print(f"  EWISER:    {ew_n:>5} instances | {ew_err:>4} errors | "
          f"{ew_err/ew_n*100:.1f}% error rate")

    # Top error patterns — GlossBERT
    gb_errors = gb_word[~gb_word['correct']]
    if len(gb_errors) > 0:
        print(f"\n  GlossBERT top error patterns:")
        top = (gb_errors
               .groupby(['gold_sense', 'glossbert_prediction'])
               .size().reset_index(name='n').nlargest(5, 'n'))
        for i, r in top.iterrows():
            print(f"    {r['gold_sense'][:38]:<40}"
                  f" → {r['glossbert_prediction'][:38]:<40}  n={r['n']}")

    # Top error patterns - EWISER
    ew_errors = ew_word[~ew_word['correct']]
    if len(ew_errors) > 0:
        print(f"\n  EWISER top error patterns:")
        top2 = (ew_errors
                .groupby(['gold_sense', 'predicted_sense_key'])
                .size().reset_index(name='n').nlargest(5, 'n'))
        for i, r in top2.iterrows():
            print(f"    {r['gold_sense'][:38]:<40}"
                  f" -> {r['predicted_sense_key'][:38]:<40}  n={r['n']}")

# Encoding bug for ONE
print("ENCODING BUG CHECK for ONE (and TWO)")
for word in ['one', 'two']:
    gb_w_err = gb[(gb['target_word'] == word) & (~gb['correct'])]
    bug = ((gb_w_err['gold_sense'].str.contains('%3:', na=False)) &
           (gb_w_err['glossbert_prediction'].str.contains('%5:', na=False))).sum()
    print(f"{word.upper()}: total errors={len(gb_w_err)} | "
          f"ss_type 3→5 bug={bug} ({bug/len(gb_w_err)*100:.0f}%) | "
          f"genuine={len(gb_w_err)-bug}")

print("SAMPLE MAN ERROR SENTENCES (how flowchart questions were designed):")
print("(Reading these sentences showed the pattern → wrote Q3 for generic use)")
gb_man_err = gb[(gb['target_word'] == 'man') & (~gb['correct']) &
                (gb['gold_sense'].str.contains('05:00', na=False)) &
                (gb['glossbert_prediction'].str.contains('18:04', na=False))]
for i, r in gb_man_err.head(5).iterrows():
    print(f"  • {r['sentence'][:100]}")

print("\nSAMPLE UNITED_STATES ERROR SENTENCES:")
ew_us_err = ew[(ew['lemma'] == 'united_states') & (~ew['correct'])]
for _, r in ew_us_err.head(5).iterrows():
    print(f"  • {r['sentence'][:100]}")

print("Flowchart classification (A / B / C)")

# MAN flowchart
GENERIC_MAN = [
    'contemporary man', 'modern man', 'biblical man', 'ancient man',
    'god created man', "man 's", 'man has ', 'man is ',
    'the nature of man', 'idea of man', 'man was made'
]

def apply_man_flowchart(gold, pred, sentence):
    sl    = sentence.lower()
    g_lex = get_lex(gold)
    p_lex = get_lex(pred)
    # Q3: generic bare-noun use -> both individual and humanity valid -> C
    if any(p in sl for p in GENERIC_MAN) and g_lex in ('05','18') and p_lex in ('05','18'):
        return ('C', 'Q3 YES: Generic bare-noun "man" without article. '
                     'Simultaneously valid as individual (05:00) and humanity (18:00). '
                     'Flowchart cannot decide - WordNet granularity problem.')
    # Same lex file -> fine-grained confusion -> C
    if g_lex and p_lex and g_lex == p_lex:
        return ('C', f'Same lex file (lex-{g_lex}) - senses too close to '
                     f'separate from text. Flowchart cannot decide between sub-senses.')
    # Cross-lex error -> genuine model error -> A
    return ('A', f'Cross-lex error: gold lex-{g_lex} vs predicted lex-{p_lex}. '
                 f'Model predicted the wrong semantic domain.')

gb_man = gb[(gb['target_word'] == 'man') & (~gb['correct'])].copy()
ew_man = ew[(ew['lemma'] == 'man')       & (~ew['correct'])].copy()

gb_man[['category', 'reason']] = pd.DataFrame(
    gb_man.apply(lambda r: apply_man_flowchart(
        r['gold_sense'], r['glossbert_prediction'], r['sentence']), axis=1).tolist(),
    index=gb_man.index)

ew_man[['category', 'reason']] = pd.DataFrame(
    ew_man.apply(lambda r: apply_man_flowchart(
        r['gold_sense'], r['predicted_sense_key'], r['sentence']), axis=1).tolist(),
    index=ew_man.index)

print(f"\nMAN — GlossBERT errors: {gb_man['category'].value_counts().to_dict()}")
print(f"MAN — EWISER errors:    {ew_man['category'].value_counts().to_dict()}")

# UNITED STATES flowchart
GOV_SIGNALS = [
    'appropriated', 'arms policy', 'nuclear', 'congress', 'senate',
    'military', 'enacted', 'legislation', 'defense', 'foreign policy',
    'has exploded', 'security'
]

def apply_us_flowchart(sentence):
    sl = sentence.lower()
    # Can you add "government" without changing meaning?
    is_gov = any(s in sl for s in GOV_SIGNALS)
    if is_gov:
        return ('B', 'Government/institutional action (supervisor test: can add '
                     '"government" without changing meaning). EWISER predicted '
                     '15:00 (government) correctly. SemCor gold 14:00 (country) '
                     'is wrong for this context.')
    return ('C', 'Country vs government indistinguishable (supervisor test: '
                 'adding "government" changes meaning). Neither sense is '
                 'exclusively signalled. Flowchart cannot decide.')

ew_us = ew[(ew['lemma'] == 'united_states') & (~ew['correct'])].copy()
ew_us['category'], ew_us['reason'] = zip(
    *ew_us['sentence'].apply(apply_us_flowchart))

print(f"\nUNITED_STATES — EWISER errors: {ew_us['category'].value_counts().to_dict()}")

# ONE/TWO classification
print("\nONE/TWO classification:")
for word in ['one', 'two']:
    gb_w = gb[(gb['target_word'] == word) & (~gb['correct'])]
    ew_w = ew[(ew['lemma'] == word)       & (~ew['correct'])]

    # GlossBERT: check encoding bug
    gb_bug     = ((gb_w['gold_sense'].str.contains('%3:', na=False)) &
                  (gb_w['glossbert_prediction'].str.contains('%5:', na=False))).sum()
    gb_genuine = len(gb_w) - gb_bug

    # EWISER: classify genuine errors
    ew_b = ew_w[ew_w['gold_sense'].str.contains('extraordinary', na=False)]
    ew_c = ew_w[ew_w['gold_sense'].str.contains('same:00', na=False)]
    ew_a = len(ew_w) - len(ew_b) - len(ew_c)

    print(f"\n  {word.upper()}:")
    print(f"    GlossBERT: ENCODING BUG = {gb_bug} ({gb_bug/len(gb_w)*100:.0f}%)  "
          f"|  GENUINE = {gb_genuine}")
    print(f"    EWISER: B (annotation error, extraordinary sense) = {len(ew_b)}  "
          f"|  C (same/borderline) = {len(ew_c)}  |  A (genuine) = {ew_a}")

print("POS filtering for EWISER")

# Step 1: Map SemCor Penn Treebank POS tags to simple codes
penn_to_simple = {
    'NN':'n', 'NNS':'n', 'NNP':'n', 'NNPS':'n',    # nouns
    'VB':'v', 'VBD':'v', 'VBZ':'v', 'VBP':'v',      # verbs
    'VBG':'v', 'VBN':'v',                             # verbs (continued)
    'JJ':'a', 'JJR':'a', 'JJS':'a',                  # adjectives
    'RB':'r', 'RBR':'r', 'RBS':'r'                    # adverbs
}
ew['gold_pos'] = ew['pos'].map(penn_to_simple)

# Step 2: Extract POS from predicted sense key
# ss_type digit after % tells us: 1=noun, 2=verb, 3/5=adjective, 4=adverb
def get_pos_from_sk(sk):
    m = re.match(r'[^%]+%(\d):', str(sk))
    if m:
        return {'1':'n', '2':'v', '3':'a', '4':'r', '5':'a'}.get(m.group(1))
    return None

ew['pred_pos']    = ew['predicted_sense_key'].apply(get_pos_from_sk)
ew['corr_before'] = (ew['predicted_sense_key'] == ew['gold_sense'])

acc_before    = ew['corr_before'].mean() * 100
n_corr_before = ew['corr_before'].sum()
print(f"\nAccuracy BEFORE filter: {acc_before:.2f}%  ({n_corr_before:,} correct)")

# Step 3: Identify cross-POS errors
cross_mask = (
    ~ew['corr_before'] &                            # prediction was wrong
    ew['gold_pos'].notna() &                        # gold POS known
    ew['pred_pos'].notna() &                        # predicted POS known
    (ew['gold_pos'] != ew['pred_pos'])              # POS mismatch!
)
print(f"Cross-POS errors found: {cross_mask.sum():,}  "
      f"({cross_mask.sum()/len(ew)*100:.1f}% of all instances)")

# Step 4: Build MFS dictionary - most frequent gold sense per (lemma, POS)
mfs_dict = {}
for (lemma, pos), group in ew.groupby(['lemma', 'gold_pos']):
    freq = group['gold_sense'].value_counts()
    if len(freq) > 0:
        mfs_dict[(lemma, pos)] = freq.index[0]
print(f"MFS dictionary: {len(mfs_dict):,} (lemma, POS) combinations")

# Step 5: Apply replacement
ew_fixed   = ew.copy()
fix_count  = 0
for idx in ew[cross_mask].index:
    replacement = mfs_dict.get((ew.at[idx, 'lemma'], ew.at[idx, 'gold_pos']))
    if replacement:
        ew_fixed.at[idx, 'predicted_sense_key'] = replacement
        fix_count += 1
print(f"Replacements applied: {fix_count:,}")

# Step 6: Compute new accuracy
ew_fixed['corr_after'] = (ew_fixed['predicted_sense_key'] == ew_fixed['gold_sense'])
acc_after    = ew_fixed['corr_after'].mean() * 100
n_corr_after = ew_fixed['corr_after'].sum()
new_correct  = n_corr_after - n_corr_before

print(f"\nAccuracy AFTER filter: {acc_after:.2f}%  ({n_corr_after:,} correct)")
print(f"Improvement:           +{acc_after - acc_before:.2f} pp  (+{new_correct:,} new correct)")
print(f"Fixed - correct:       {new_correct:,}/{fix_count:,} = {new_correct/fix_count*100:.0f}%")
print(f"Fixed - still wrong:   {fix_count - new_correct:,} (MFS was not the gold sense)")

# Save output
out = ew_fixed[['sentence', 'target_word', 'lemma', 'pos',
                'gold_sense', 'predicted_sense_key',
                'corr_before', 'corr_after']].copy()
out['pos_filter_applied'] = cross_mask.apply(lambda x: 'YES' if x else 'no')
OUT_POS = os.path.join(DRIVE_FOLDER, 'ewiser_pos_filtered.csv')
out.to_csv(OUT_POS, index=False)
print(f"\nSaved: {OUT_POS}")
print(f"  Rows: {len(out):,} | Fixed rows: {(out['pos_filter_applied']=='YES').sum():,}")

print("Flowchart limitations (Category C breakdown per word)")

SHARED_200_CSV = os.path.join(DRIVE_FOLDER, 'Example_200_SharedSentences.csv')

if not os.path.exists(SHARED_200_CSV):
    print(f" File not found: {SHARED_200_CSV}")
else:
    df200 = pd.read_csv(SHARED_200_CSV)
    print(f"\nLoaded: {len(df200)} rows")
    print(f"Overall: A={( df200['category']=='A').sum()}, "
          f"B={(df200['category']=='B').sum()}, "
          f"C={(df200['category']=='C').sum()}")
    print()
    print(f"{'Word':<8} {'Total':>7} {'Cat A':>8} {'Cat B':>8} {'Cat C':>8} {'C%':>8}")
    for word in ['make', 'have', 'do']:
        sub = df200[df200['word'] == word]
        a = (sub['category'] == 'A').sum()
        b = (sub['category'] == 'B').sum()
        c = (sub['category'] == 'C').sum()
        print(f"{word:<8} {len(sub):>7} {a:>8} {b:>8} {c:>8} {c/len(sub)*100:>7.0f}%")

    overall_c = (df200['category'] == 'C').sum()
    print(f"\nOverall Category C: {overall_c} / {len(df200)} = "
          f"{overall_c/len(df200)*100:.1f}%")
    print("(This is the upper bound on granularity-driven errors)")
```

```
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import re
import os

DRIVE_FOLDER  = '/content/drive/MyDrive/EWISER'   #  change if needed
GB_FILE       = os.path.join(DRIVE_FOLDER, 'glossbert_full.csv')
EW_FILE       = os.path.join(DRIVE_FOLDER, 'ewiser_full_mfs_with_sensekey.csv')
GB_OUT_FILE   = os.path.join(DRIVE_FOLDER, 'glossbert_encoding_bug_fixed.csv')
EW_OUT_FILE   = os.path.join(DRIVE_FOLDER, 'ewiser_encoding_bug_fixed.csv')

for f in [GB_FILE, EW_FILE]:
    print(f"{'found' if os.path.exists(f) else 'NOT FOUND'} {f}")

print("\nLoading GlossBERT predictions...")
gb = pd.read_csv(GB_FILE)
gb['correct'] = gb['glossbert_prediction'] == gb['gold_sense']
print(f"  Rows: {len(gb):,} | Raw accuracy: {gb['correct'].mean()*100:.2f}%")

print("Loading EWISER predictions...")
ew = pd.read_csv(EW_FILE)
ew['correct'] = ew['predicted_sense_key'] == ew['gold_sense']
print(f"  Rows: {len(ew):,} | Raw accuracy: {ew['correct'].mean()*100:.2f}%")

def has_encoding_bug(gold, pred):
    """
    Detects the ss_type 3 vs 5 encoding format mismatch.

    The format mismatch occurs when:
      - SemCor gold uses ss_type=3 (Princeton WN 1993 format)
      - Model prediction uses ss_type=5 (OEWN 2025 format)
      - Every other field (lemma, lex_filenum, lex_id) is identical

    These two sense keys refer to the EXACT same sense — only the format
    digit after % differs. They should be treated as correct predictions.

    Returns: True if this is an encoding bug (not a real error)
    """
    if pd.isna(gold) or pd.isna(pred):
        return False
    # Normalise gold from ss_type=3 to ss_type=5 format and compare
    gold_normalised = re.sub(r'%3:', '%5:', str(gold))
    return gold_normalised == str(pred)

print("ENCODING BUG ANALYSIS — BOTH MODELS")

# Mark encoding bugs in GlossBERT
gb['is_encoding_bug'] = gb.apply(
    lambda r: (not r['correct']) and has_encoding_bug(
        r['gold_sense'], r['glossbert_prediction']),
    axis=1
)
gb['correct_after_bug_fix'] = gb['correct'] | gb['is_encoding_bug']

# Mark encoding bugs in EWISER
ew['is_encoding_bug'] = ew.apply(
    lambda r: (not r['correct']) and has_encoding_bug(
        r['gold_sense'], r['predicted_sense_key']),
    axis=1
)
ew['correct_after_bug_fix'] = ew['correct'] | ew['is_encoding_bug']

# Results
gb_bug = gb['is_encoding_bug'].sum()
ew_bug = ew['is_encoding_bug'].sum()
gb_err = (~gb['correct']).sum()
ew_err = (~ew['correct']).sum()
gb_raw = gb['correct'].mean() * 100
gb_fix = gb['correct_after_bug_fix'].mean() * 100
ew_raw = ew['correct'].mean() * 100
ew_fix = ew['correct_after_bug_fix'].mean() * 100

print(f"\nGlossBERT:")
print(f"  Total errors:       {gb_err:,}")
print(f"  Encoding bugs:      {gb_bug:,} ({gb_bug/gb_err*100:.1f}% of all errors)")
print(f"  Raw accuracy:       {gb_raw:.2f}%  ({gb['correct'].sum():,} correct)")
print(f"  Corrected accuracy: {gb_fix:.2f}%  ({gb['correct_after_bug_fix'].sum():,} correct)")
print(f"  Improvement:        +{gb_fix-gb_raw:.2f} percentage points")

print(f"\nEWISER:")
print(f"  Total errors:       {ew_err:,}")
print(f"  Encoding bugs:      {ew_bug:,} ({ew_bug/ew_err*100:.1f}% of all errors)")
print(f"  Raw accuracy:       {ew_raw:.2f}%  ({ew['correct'].sum():,} correct)")
print(f"  Corrected accuracy: {ew_fix:.2f}%  ({ew['correct_after_bug_fix'].sum():,} correct)")
print(f"  Improvement:        +{ew_fix-ew_raw:.2f} percentage points")

print(f"  EWISER uses OEWN 2025 for its graph-convolution candidate scoring,")
print(f"  so its predictions also carry ss_type=5 for adjective satellite senses.")
print(f"  SemCor gold uses ss_type=3 (Princeton WN 1993).")
print(f"  Examples of EWISER encoding bugs:")
ew_bug_rows = ew[ew['is_encoding_bug']].head(4)
for i, r in ew_bug_rows.iterrows():
    print(f"    {r['lemma']:<15} gold={r['gold_sense'][:35]:<35} "
          f"pred={r['predicted_sense_key'][:35]}")

# Per-word encoding bug counts
print("\n\nWords most affected by encoding bug (both models):")
print(f"{'Word':<15} {'GB bugs':>8} {'EW bugs':>8} {'GB raw%':>9} {'GB corrected%':>14}")
print("-"*60)
all_words = set(gb['target_word'].unique()) & set(ew['lemma'].unique())
word_stats = []
for word in all_words:
    gb_w = gb[gb['target_word']==word]
    ew_w = ew[ew['lemma']==word]
    if len(gb_w) < 20: continue
    gb_bugs = gb_w['is_encoding_bug'].sum()
    ew_bugs = ew_w['is_encoding_bug'].sum()
    if gb_bugs > 50:
        gb_r = (~gb_w['correct']).mean()*100
        gb_c = (~gb_w['correct_after_bug_fix'].isin([False])).mean()*100
        word_stats.append({'word':word,'gb_bugs':gb_bugs,'ew_bugs':ew_bugs,
                          'gb_raw':gb_r,'gb_corr':gb_w['correct_after_bug_fix'].mean()*100})

word_stats_df = pd.DataFrame(word_stats).sort_values('gb_bugs',ascending=False)
for i, r in word_stats_df.head(12).iterrows():
    print(f"{r['word']:<15} {r['gb_bugs']:>8,} {r['ew_bugs']:>8,} "
          f"{r['gb_raw']:>8.1f}% {r['gb_corr']:>13.1f}%")

# GlossBERT corrected file
gb_out = gb[['sentence','target_word','gold_sense','glossbert_prediction',
             'correct','is_encoding_bug','correct_after_bug_fix']].copy()
gb_out.to_csv(GB_OUT_FILE, index=False)
print(f"Saved: {GB_OUT_FILE}")
print(f"  Rows: {len(gb_out):,}")
print(f"  Encoding bugs: {gb_out['is_encoding_bug'].sum():,}")

# EWISER corrected file
ew_out = ew[['sentence','target_word','lemma','pos','gold_sense','predicted_sense_key',
             'correct','is_encoding_bug','correct_after_bug_fix']].copy()
ew_out.to_csv(EW_OUT_FILE, index=False)
print(f"Saved: {EW_OUT_FILE}")
print(f"  Rows: {len(ew_out):,}")
print(f"  Encoding bugs: {ew_out['is_encoding_bug'].sum():,}")

print("SUMMARY TABLE FOR PAPER (Table to add)")
print(f"{'System':<25} {'Raw Accuracy':>14} {'Bug-Corrected':>14} {'Improvement':>12}")
print(f"{'GlossBERT':<25} {gb_raw:>13.2f}% {gb_fix:>13.2f}% {gb_fix-gb_raw:>11.2f} pp")
print(f"{'EWISER + MFS':<25} {ew_raw:>13.2f}% {ew_fix:>13.2f}% {ew_fix-ew_raw:>11.2f} pp")
```

```
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd, re, os

DRIVE_FOLDER  = '/content/drive/MyDrive/EWISER'
GB_FILE       = os.path.join(DRIVE_FOLDER, 'glossbert_full.csv')
EW_FILE       = os.path.join(DRIVE_FOLDER, 'ewiser_full_mfs_with_sensekey.csv')
OUT_FILE      = os.path.join(DRIVE_FOLDER, 'adjective_good_classified.csv')

print("Loading prediction files...")
gb = pd.read_csv(GB_FILE)
ew = pd.read_csv(EW_FILE)
gb['correct'] = gb['glossbert_prediction'] == gb['gold_sense']
ew['correct']  = ew['predicted_sense_key']  == ew['gold_sense']

# Encoding bug detection
def has_encoding_bug(gold, pred):
    if pd.isna(gold) or pd.isna(pred): return False
    return re.sub(r'%3:', '%5:', str(gold)) == str(pred)

gb['is_encoding_bug'] = gb.apply(
    lambda r: (not r['correct']) and has_encoding_bug(
        r['gold_sense'], r['glossbert_prediction']), axis=1)
ew['is_encoding_bug'] = ew.apply(
    lambda r: (not r['correct']) and has_encoding_bug(
        r['gold_sense'], r['predicted_sense_key']), axis=1)

print(f"GlossBERT: {len(gb):,} rows | Accuracy: {gb['correct'].mean()*100:.2f}%")
print(f"EWISER:    {len(ew):,} rows | Accuracy: {ew['correct'].mean()*100:.2f}%")

# Extract "good" instances
print("ADJECTIVE: 'good' - Instance counts and error rates")

TARGET = 'good'
gb_good = gb[gb['target_word'] == TARGET].copy()
ew_good = ew[ew['lemma']        == TARGET].copy()

gb_good_err     = gb_good[~gb_good['correct']]
ew_good_err     = ew_good[~ew_good['correct']]
gb_good_genuine = gb_good[~gb_good['correct'] & ~gb_good['is_encoding_bug']]
ew_good_genuine = ew_good[~ew_good['correct'] & ~ew_good['is_encoding_bug']]

print(f"\nGlossBERT:")
print(f"  Total instances:   {len(gb_good):,}")
print(f"  Total errors:      {len(gb_good_err):,} ({len(gb_good_err)/len(gb_good)*100:.1f}%)")
print(f"  Encoding bugs:     {gb_good['is_encoding_bug'].sum():,}")
print(f"  Genuine errors:    {len(gb_good_genuine):,} ({len(gb_good_genuine)/len(gb_good)*100:.1f}%)")

print(f"\nEWISER:")
print(f"  Total instances:   {len(ew_good):,}")
print(f"  Total errors:      {len(ew_good_err):,} ({len(ew_good_err)/len(ew_good)*100:.1f}%)")
print(f"  Encoding bugs:     {ew_good['is_encoding_bug'].sum():,}")
print(f"  Genuine errors:    {len(ew_good_genuine):,} ({len(ew_good_genuine)/len(ew_good)*100:.1f}%)")

print("\nSense distribution in SemCor (all instances):")
sense_dist = pd.concat([gb_good['gold_sense'], ew_good['gold_sense']]).value_counts()
for sense, count in sense_dist.head(12).items():
    print(f"  {sense:<50} {count:>4} instances")

# Top error patterns
print("TOP ERROR PATTERNS (excluding encoding bugs)")
print("\nGlossBERT genuine errors:")
for (g,p),n in gb_good_genuine.groupby(
        ['gold_sense','glossbert_prediction']).size().nlargest(8).items():
    print(f"  {g[:45]:<45} -> {p[:35]:<35} n={n}")

print("\nEWISER genuine errors:")
for (g,p),n in ew_good_genuine.groupby(
        ['gold_sense','predicted_sense_key']).size().nlargest(8).items():
    print(f"  {g[:45]:<45} -> {p[:35]:<35} n={n}")

# Disambiguation flowchart for "good"
# Built from reading actual SemCor sense keys and their definitions in OEWN 2025
print("\n" + "="*55)
print("DISAMBIGUATION FLOWCHART for 'good'")
print("="*55)
print("""
Flowchart questions (built from actual SemCor sense keys + OEWN 2025 definitions):

Q1. Is the POS tag an adverb (RB) and does context mean "well/thoroughly"?
    - YES - good%4:02:00:: or good%4:02:01:: (adverb senses)
    - NO  - Q2

Q2. Is "good" used as a NOUN? (subject or object position, no head noun)
    (e.g. "for the good of", "for good", "do good")
    - YES - good%1:07:00:: (benefit/advantage) or good%1:07:01:: (moral excellence)
    - NO  - Q3

Q3. Is the context about moral virtue or character?
    Test: can "virtuous", "morally upright" be substituted?
    (e.g. "a good man", "good deeds", "good intentions")
    - YES - good%3:00:01:: (morally admirable — MFS)
    - NO  - Q4

Q4. Is the context about health, diet, or benefit?
    Test: can "beneficial for", "healthy for" be substituted?
    (e.g. "good for your health", "good for you")
    - YES - good%3:00:02:: (beneficial/advantageous)
    - NO  - Q5

Q5. Is the sense about thoroughness or completeness?
    Test: can "thorough", "considerable", "a good beating/talking-to" apply?
    (e.g. "gave him a good talking-to", "a good many")
    - YES - good%3:00:00:ample:00 (ample/thorough/complete)
    - NO  - Q6

Q6. Is reputation or creditworthiness the focus?
    (e.g. "good credit", "a good firm", "good name")
    - YES - good%3:00:00:reputable:00 (creditable/of good reputation)
    - NO  - CATEGORY C: general high-quality sense (good%3:00:01:: and
             good%3:00:00:advantageous:: both valid — senses too close to separate)
""")

# Apply flowchart to classify errors
def get_lex(sk):
    m = re.match(r'[^%]+%(\d):(\d{2}):', str(sk))
    return (m.group(1), m.group(2)) if m else (None, None)

def classify_good(gold, pred, sent):
    """
    Apply the 'good' flowchart to classify one error instance.
    Returns (category, reason)
    """
    sent_l = sent.lower()

    # Encoding bug check first
    if has_encoding_bug(gold, pred):
        return ('ENCODING BUG',
                f'ss_type 3→5 format mismatch. '
                f'Gold: {gold[:35]} | Pred: {pred[:35]}. Not a real error.')

    g_ss, g_lex = get_lex(gold)
    p_ss, p_lex = get_lex(pred)

    # Q1: Adverb senses
    if g_ss == '4' or p_ss == '4':
        if g_ss != p_ss:
            return ('A', f'Cross-POS error: gold ss_type={g_ss} vs pred ss_type={p_ss}. '
                        f'Adverb/adjective boundary. Gold: {gold[:30]}')
        return ('C', 'Both adverb senses (4:02:00 vs 4:02:01) — '
                    'fine-grained within adverb lex file. Flowchart cannot separate.')

    # Q2: Noun senses
    if g_ss == '1' and p_ss != '1':
        return ('A', f'Cross-POS error: gold is noun (ss_type=1), pred is adjective/adverb. '
                    f'Gold: {gold[:30]} | Pred: {pred[:30]}')
    if g_ss == '1' and p_ss == '1':
        return ('C', 'Both noun senses (1:07:00 vs 1:07:01) — '
                    'benefit vs moral excellence, fine-grained within lex-07.')

    # Q3: Moral context
    moral_signals = ['good man','good men','good person','good people',
                     'good heart','good character','good intentions','good deeds',
                     'good christian','good soul','good will','good faith']
    if any(sig in sent_l for sig in moral_signals):
        if '00:01' in gold:
            return ('A', f'Moral virtue context (Q3). Flowchart -> good%3:00:01. '
                        f'Model predicted {pred[:30]}. Genuine error.')
        if '00:01' in pred and '00:01' not in gold:
            return ('B', f'Moral context but gold is not 00:01 (morally admirable). '
                        f'Gold: {gold[:30]} — may be annotation error. Model predicts 00:01 correctly.')
        return ('C', f'Moral context but Q3 ambiguous: '
                    f'gold={gold[:25]} vs pred={pred[:25]}. Both senses valid.')

    # Q4: Health/benefit context
    health_signals = ['good for', 'do you good', 'do him good', 'do her good',
                      'good medicine', 'good for the', 'beneficial']
    if any(sig in sent_l for sig in health_signals):
        if '00:02' in gold:
            return ('A', f'Health/benefit context (Q4). Flowchart -> good%3:00:02. '
                        f'Model predicted {pred[:30]}. Genuine error.')
        return ('C', f'Q4 borderline: beneficial vs general quality. '
                    f'Gold: {gold[:25]} | Pred: {pred[:25]}.')

    # Q5: Thoroughness/ample context
    ample_signals = ['a good many', 'a good deal', 'a good while', 'a good bit',
                     'a good talking', 'a good number', 'a good long',
                     'a good time', 'good and', 'good and ready']
    if any(sig in sent_l for sig in ample_signals):
        if 'ample' in gold:
            return ('A', f'Thoroughness/ample context (Q5). Flowchart -> ample sense. '
                        f'Model predicted {pred[:30]}. Genuine error.')
        return ('C', f'Q5 borderline: thorough vs general quality. '
                    f'Gold: {gold[:25]} | Pred: {pred[:25]}.')

    # Q6: Reputation context
    rep_signals = ['good name', 'good credit', 'good reputation', 'good firm',
                   'good family', 'good standing', 'a good school']
    if any(sig in sent_l for sig in rep_signals):
        if 'reputable' in gold:
            return ('A', f'Reputation context (Q6). Flowchart → reputable sense. '
                        f'Model predicted {pred[:30]}. Genuine error.')
        return ('C', f'Q6 borderline: reputable vs generally good. Both valid.')

    # Same lex file -> Category C (senses too fine-grained)
    if g_lex and p_lex and g_lex == p_lex and g_ss == p_ss:
        return ('C', f'Same lex file (ss={g_ss}, lex={g_lex}). '
                    f'Gold ({gold[:25]}) and pred ({pred[:25]}) are both "good" '
                    f'sub-senses in the same semantic domain. '
                    f'WordNet sub-sense distinctions not recoverable from text.')

    # Cross-lex or cross-ss -> Category A
    return ('A', f'Genuine model error: gold (ss={g_ss},lex={g_lex}) '
                f'vs pred (ss={p_ss},lex={p_lex}). '
                f'Model predicted different sense class. '
                f'Gold: {gold[:25]} | Pred: {pred[:25]}')

print("Applying flowchart to all genuine errors...")

gb_good_err2 = gb_good_err.copy()
gb_good_err2[['category','reason']] = pd.DataFrame(
    gb_good_err2.apply(
        lambda r: classify_good(
            r['gold_sense'], r['glossbert_prediction'], r['sentence']),
        axis=1).tolist(), index=gb_good_err2.index)

ew_good_err2 = ew_good_err.copy()
ew_good_err2[['category','reason']] = pd.DataFrame(
    ew_good_err2.apply(
        lambda r: classify_good(
            r['gold_sense'], r['predicted_sense_key'], r['sentence']),
        axis=1).tolist(), index=ew_good_err2.index)

print(f"\nGlossBERT 'good' classification:")
print(f"  {gb_good_err2['category'].value_counts().to_dict()}")
print(f"\nEWISER 'good' classification:")
print(f"  {ew_good_err2['category'].value_counts().to_dict()}")

# Show sample sentences for each category
print("SAMPLE SENTENCES PER CATEGORY")
for cat in ['A','B','C','ENCODING BUG']:
    gb_sub = gb_good_err2[gb_good_err2['category']==cat]
    ew_sub = ew_good_err2[ew_good_err2['category']==cat]
    if len(gb_sub)==0 and len(ew_sub)==0: continue
    print(f"\n--- Category {cat} ---")
    for i, r in gb_sub.head(2).iterrows():
        print(f"  [GB] {r['sentence'][:90]}")
        print(f"       Gold: {r['gold_sense']} | Pred: {r['glossbert_prediction']}")
        print(f"       Reason: {r['reason'][:80]}")
    for i, r in ew_sub.head(2).iterrows():
        print(f"  [EW] {r['sentence'][:90]}")
        print(f"       Gold: {r['gold_sense']} | Pred: {r['predicted_sense_key']}")
        print(f"       Reason: {r['reason'][:80]}")

print("SAVING EVIDENCE FILE")

gb_out = gb_good_err2[['sentence','target_word','gold_sense',
                        'glossbert_prediction','category','reason']].copy()
gb_out['model'] = 'GlossBERT'
gb_out.rename(columns={'glossbert_prediction':'prediction'}, inplace=True)

ew_out = ew_good_err2[['sentence','target_word','gold_sense',
                        'predicted_sense_key','category','reason']].copy()
ew_out['model'] = 'EWISER'
ew_out.rename(columns={'predicted_sense_key':'prediction'}, inplace=True)

combined = pd.concat([gb_out, ew_out]).reset_index(drop=True)
combined.insert(0, 'id', range(1, len(combined)+1))
combined.to_csv(OUT_FILE, index=False)

print(f"Saved: {OUT_FILE}")
print(f"Total rows: {len(combined)}")
print(f"\nFinal summary (both models):")
print(combined.groupby(['model','category']).size().unstack(fill_value=0))

# Results to use in paper
print("NUMBERS FOR THE PAPER")
gb_cats = gb_good_err2['category'].value_counts()
ew_cats = ew_good_err2['category'].value_counts()
total_gb = len(gb_good_err2)
total_ew = len(ew_good_err2)
print(f"Adjective 'good': GB={total_gb} errors (total incl. bugs), "
      f"EW={total_ew} errors (total incl. bugs)")
print(f"\nGlossBERT classification:")
for cat in ['A','B','C','ENCODING BUG']:
    n = gb_cats.get(cat,0)
    print(f"  {cat}: {n} ({n/total_gb*100:.0f}%)")
print(f"\nEWISER classification:")
for cat in ['A','B','C','ENCODING BUG']:
    n = ew_cats.get(cat,0)
    print(f"  {cat}: {n} ({n/total_ew*100:.0f}%)")
```

```
# Mount Drive
from google.colab import drive
drive.mount('/content/drive')

# Load corrected prediction files
import pandas as pd
import re
import os

DRIVE_FOLDER = '/content/drive/MyDrive/EWISER'
GB_FILE      = os.path.join(DRIVE_FOLDER, 'glossbert_full.csv')
EW_FILE      = os.path.join(DRIVE_FOLDER, 'ewiser_full_mfs_with_sensekey.csv')
GB_FIXED     = os.path.join(DRIVE_FOLDER, 'glossbert_encoding_bug_fixed.csv')
EW_FIXED     = os.path.join(DRIVE_FOLDER, 'ewiser_encoding_bug_fixed.csv')

# Load original files
print("Loading original prediction files...")
gb = pd.read_csv(GB_FILE)
ew = pd.read_csv(EW_FILE)
gb['correct'] = gb['glossbert_prediction'] == gb['gold_sense']
ew['correct']  = ew['predicted_sense_key']  == ew['gold_sense']

# Load or recompute bug-corrected columns
def has_encoding_bug(gold, pred):
    if pd.isna(gold) or pd.isna(pred): return False
    return re.sub(r'%3:', '%5:', str(gold)) == str(pred)

if os.path.exists(GB_FIXED) and os.path.exists(EW_FIXED):
    print("Loading bug-corrected files...")
    gb_fix = pd.read_csv(GB_FIXED)
    ew_fix = pd.read_csv(EW_FIXED)
    gb['correct_fixed'] = gb_fix['correct_after_bug_fix']
    ew['correct_fixed'] = ew_fix['correct_after_bug_fix']
    gb['is_bug'] = gb_fix['is_encoding_bug']
    ew['is_bug'] = ew_fix['is_encoding_bug']
else:
    print("Bug-fixed files not found — recomputing encoding bug flag...")
    gb['is_bug'] = gb.apply(
        lambda r: (not r['correct']) and has_encoding_bug(
            r['gold_sense'], r['glossbert_prediction']), axis=1)
    ew['is_bug'] = ew.apply(
        lambda r: (not r['correct']) and has_encoding_bug(
            r['gold_sense'], r['predicted_sense_key']), axis=1)
    gb['correct_fixed'] = gb['correct'] | gb['is_bug']
    ew['correct_fixed'] = ew['correct'] | ew['is_bug']

gb_bug = gb['is_bug'].sum()
ew_bug = ew['is_bug'].sum()
print(f"GlossBERT encoding bugs: {gb_bug:,}")
print(f"EWISER    encoding bugs: {ew_bug:,}")

# Overall accuracy (before and after)
print("Overall Accuracy - Before vs After Bug Fix")

gb_raw  = gb['correct'].mean() * 100
gb_corr = gb['correct_fixed'].mean() * 100
ew_raw  = ew['correct'].mean() * 100
ew_corr = ew['correct_fixed'].mean() * 100

print(f"\n{'System':<30} {'Raw accuracy':>14} {'Bug-corrected':>14} {'Change':>10}")
print(f"{'GlossBERT':<30} {gb_raw:>13.2f}% {gb_corr:>13.2f}% {gb_corr-gb_raw:>8.2f} pp")
print(f"{'EWISER + MFS':<30} {ew_raw:>13.2f}% {ew_corr:>13.2f}% {ew_corr-ew_raw:>8.2f} pp")
print(f"\nNote: GliteTech LLM (~80%) also represents a corrected-annotation baseline")
print(f"All three systems converge at 80-91% after removing format artefacts")

# Same-lex error % per word
print("Same-Lex Error % - Bug-Corrected (Light Verbs)")

def get_lex(sk):
    m = re.match(r'[^%]+%\d:(\d{2}):', str(sk))
    return m.group(1) if m else None

print(f"\n{'Word':<10} {'GB err (raw)':>12} {'GB err (fix)':>12} "
      f"{'EW err (raw)':>12} {'EW err (fix)':>12} {'GB lex%':>9} {'EW lex%':>9}")

LIGHT_VERBS = ['make','have','do','find','give','see','be']
for word in LIGHT_VERBS:
    gb_w = gb[gb['target_word']==word]
    ew_w = ew[ew['lemma']==word]

    gb_err_raw  = (~gb_w['correct']).sum()
    gb_err_fix  = (~gb_w['correct_fixed']).sum()
    ew_err_raw  = (~ew_w['correct']).sum()
    ew_err_fix  = (~ew_w['correct_fixed']).sum()

    # Same-lex % computed on FIXED errors only
    gb_fixed_errors = gb_w[~gb_w['correct_fixed']].copy()
    ew_fixed_errors = ew_w[~ew_w['correct_fixed']].copy()
    gb_fixed_errors['gl'] = gb_fixed_errors['gold_sense'].apply(get_lex)
    gb_fixed_errors['pl'] = gb_fixed_errors['glossbert_prediction'].apply(get_lex)
    ew_fixed_errors['gl'] = ew_fixed_errors['gold_sense'].apply(get_lex)
    ew_fixed_errors['pl'] = ew_fixed_errors['predicted_sense_key'].apply(get_lex)

    gb_same = (gb_fixed_errors['gl']==gb_fixed_errors['pl']).mean()*100 if len(gb_fixed_errors)>0 else 0
    ew_same = (ew_fixed_errors['gl']==ew_fixed_errors['pl']).mean()*100 if len(ew_fixed_errors)>0 else 0

    print(f"{word:<10} {gb_err_raw:>12,} {gb_err_fix:>12,} "
          f"{ew_err_raw:>12,} {ew_err_fix:>12,} {gb_same:>8.1f}% {ew_same:>8.1f}%")

print(f"\n{'Word':<18} {'GB raw%':>9} {'GB fixed%':>10} "
      f"{'EW raw%':>9} {'EW fixed%':>10} {'Bug affects?'}")

for word in ['man','united_states','one','two','good']:
    gb_w = gb[gb['target_word']==word]
    ew_w = ew[ew['lemma']==word]
    if len(gb_w)==0 and len(ew_w)==0: continue

    gb_r = (~gb_w['correct']).mean()*100 if len(gb_w)>0 else 0
    gb_f = (~gb_w['correct_fixed']).mean()*100 if len(gb_w)>0 else 0
    ew_r = (~ew_w['correct']).mean()*100 if len(ew_w)>0 else 0
    ew_f = (~ew_w['correct_fixed']).mean()*100 if len(ew_w)>0 else 0

    affected = 'YES' if abs(gb_f-gb_r)>1 or abs(ew_f-ew_r)>1 else '- no'
    print(f"{word:<18} {gb_r:>8.1f}% {gb_f:>9.1f}% "
          f"{ew_r:>8.1f}% {ew_f:>9.1f}% {affected}")

print("\nFor ONE and TWO: massive improvement because nearly all errors were encoding bugs.")
print("For MAN and UNITED_STATES: no change (nouns use ss_type=1, unaffected).")

# Shared failure analysis
print("Shared Failure - Bug-Corrected (Both Models)")

# Build composite key and merge
gb['row_id'] = (gb['sentence'].str.strip()+'||'+
                gb['target_word'].str.strip()+'||'+
                gb['gold_sense'].str.strip())
ew['row_id'] = (ew['sentence'].str.strip()+'||'+
                ew['target_word'].str.strip()+'||'+
                ew['gold_sense'].str.strip())

merged = gb[['row_id','target_word','correct','correct_fixed']].rename(
    columns={'correct':'gb_raw','correct_fixed':'gb_fix'}).merge(
    ew[['row_id','correct','correct_fixed']].rename(
    columns={'correct':'ew_raw','correct_fixed':'ew_fix'}),
    on='row_id', how='inner')

total = len(merged)
print(f"Total matched instances: {total:,}")

# Before fix
bc_r = (merged['gb_raw']&merged['ew_raw']).sum()
bw_r = (~merged['gb_raw']&~merged['ew_raw']).sum()
og_r = (~merged['gb_raw']&merged['ew_raw']).sum()
oe_r = (merged['gb_raw']&~merged['ew_raw']).sum()

# After fix
bc_f = (merged['gb_fix']&merged['ew_fix']).sum()
bw_f = (~merged['gb_fix']&~merged['ew_fix']).sum()
og_f = (~merged['gb_fix']&merged['ew_fix']).sum()
oe_f = (merged['gb_fix']&~merged['ew_fix']).sum()

print(f"\n{'Outcome':<25} {'Before fix':>12} {'%':>7} {'After fix':>11} {'%':>7} {'Change':>9}")
print("-"*78)
print(f"{'Both models correct':<25} {bc_r:>12,} {bc_r/total*100:>6.1f}% "
      f"{bc_f:>11,} {bc_f/total*100:>6.1f}% {(bc_f-bc_r)/total*100:>+8.1f}pp")
print(f"{'Both models wrong':<25} {bw_r:>12,} {bw_r/total*100:>6.1f}% "
      f"{bw_f:>11,} {bw_f/total*100:>6.1f}% {(bw_f-bw_r)/total*100:>+8.1f}pp")
print(f"{'Only GlossBERT wrong':<25} {og_r:>12,} {og_r/total*100:>6.1f}% "
      f"{og_f:>11,} {og_f/total*100:>6.1f}% {(og_f-og_r)/total*100:>+8.1f}pp")
print(f"{'Only EWISER wrong':<25} {oe_r:>12,} {oe_r/total*100:>6.1f}% "
      f"{oe_f:>11,} {oe_f/total*100:>6.1f}% {(oe_f-oe_r)/total*100:>+8.1f}pp")

print(f"\nKey finding: 'Both wrong' drops from {bw_r/total*100:.1f}% → {bw_f/total*100:.1f}%")
print(f"  This means {bw_r-bw_f:,} of the 'both wrong' cases were encoding artefacts,")
print(f"  not genuine shared difficulty. True shared failure rate = {bw_f/total*100:.1f}%.")

# Word-level shared failure after fix
print(f"\nWord-level breakdown AFTER fix:")
print(f"{'Word':<8} {'Total':>7} {'Both✓':>8} {'Both✗':>8} "
      f"{'Only GB✗':>10} {'Only EW✗':>10} {'GB fix%':>9} {'EW fix%':>9}")
print("-"*80)

rows = []
for word in ['make','have','do','find','give','see','be','man','one','two']:
    m = merged[merged['target_word']==word]
    if len(m)<5: continue
    bc = (m['gb_fix']&m['ew_fix']).sum()
    bw = (~m['gb_fix']&~m['ew_fix']).sum()
    og = (~m['gb_fix']&m['ew_fix']).sum()
    oe = (m['gb_fix']&~m['ew_fix']).sum()
    gbe = (~m['gb_fix']).mean()*100
    ewe = (~m['ew_fix']).mean()*100
    print(f"{word:<8} {len(m):>7,} {bc:>8,} {bw:>8,} {og:>10,} "
          f"{oe:>10,} {gbe:>8.1f}% {ewe:>8.1f}%")
    rows.append({'word':word,'total':len(m),'both_correct_fixed':bc,
                 'both_wrong_fixed':bw,'only_gb_wrong_fixed':og,
                 'only_ew_wrong_fixed':oe,'gb_err_pct_fixed':round(gbe,1),
                 'ew_err_pct_fixed':round(ewe,1)})

# Save updated shared failure
out_shared = os.path.join(DRIVE_FOLDER, 'shared_failure_by_word_bugfixed.csv')
pd.DataFrame(rows).to_csv(out_shared, index=False)
print(f"\nSaved: {out_shared}")
```
